## 2. 三足步态的概念模型

下面仅演示两组交替的抽象相位，A/B 是占位组别，不代表 Jumper 实际的腿部命名、触地时间或训练策略。实际分组应查看上游任务实现及回放。

```python
for step in range(8):
    support = 'A' if step % 2 == 0 else 'B'
    swing = 'B' if support == 'A' else 'A'
    print(f'相位 {step}: 支撑组 {support}, 摆动组 {swing}')
```
