## 1. 检查环境

Python 和 Git 用于阅读与运行工具；GPU 检查仅报告当前环境，具体训练依赖以项目指南为准。

````python
import sys, platform, shutil, subprocess
print('Python:', sys.version.split()[0])
print('系统:', platform.platform())
for name in ('git', 'nvidia-smi'):
    path = shutil.which(name)
    print(name, path or '未找到')
    if name == 'nvidia-smi' and path:
        p = subprocess.run([path, '--query-gpu=name,driver_version', '--format=csv,noheader'], capture_output=True, text=True)
        print(p.stdout.strip() or p.stderr.strip())
````
